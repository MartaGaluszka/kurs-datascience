```python
from sqlalchemy import Column, Integer, String, Text, DateTime, create_engine
from sqlalchemy.orm import declarative_base

Base = declarative_base()

# Initial version of the Article model
class Article(Base):
    __tablename__ = 'articles'
    id = Column(Integer, primary_key=True)
    title = Column(String(200), nullable=False)
    content = Column(Text)
    # We will add 'published_date' later via migration!

print("✅ Initial Article model defined.")
```

```python
# Updated version of the Article model (after migration)
class Article(Base):
    __tablename__ = 'articles'

    __table_args__ = {'extend_existing': True}

    id = Column(Integer, primary_key=True)
    title = Column(String(200), nullable=False)
    content = Column(Text)
    # This was added via Alembic migration:
    published_date = Column(DateTime) 

print("✅ Article model updated with published_date.")
```
