```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Tworzymy dane (DataFrame)
data = {
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'USB Cable', 
                     'Headphones', 'Webcam', 'Hard Drive', 'RAM', 'GPU', 'CPU'],
    'price': [3500, 50, 150, 800, 25, 200, 300, 450, 400, 2500, 1800],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Cables',
                 'Audio', 'Video', 'Storage', 'Components', 'Components', 'Components']
}

df = pd.DataFrame(data)
print("✅ DataFrame ready!")
print(df.head())
```

```python
# 2. Tworzymy połączenie z bazą store.db
engine_store = create_engine('sqlite:///store.db')

# 3. Przesyłamy DataFrame do bazy danych
# 'products' to nazwa tabeli, która powstanie w pliku .db
df.to_sql('products', con=engine_store, if_exists='replace', index=False)

print("✅ Data successfully exported to store.db in 'products' table!")
```

```python
# 4. Sprawdzamy, czy możemy odczytać te dane z powrotem z bazy
df_from_db = pd.read_sql_table('products', engine_store)

print("--- Data read from SQL ---")
print(df_from_db)
```
