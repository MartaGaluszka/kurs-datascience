Uniwersalny Szablon Wykresu (Matplotlib)

```python
import matplotlib.pyplot as plt

# KROK 1: Przygotowanie danych (np. z listy lub DataFrame)
x = [1, 2, 3, 4, 5]
y = [10, 24, 33, 41, 55]

# KROK 2: Inicjalizacja "płótna" (Metoda obiektowa - najlepsza praktyka)
fig, ax = plt.subplots(figsize=(6, 3)) 

# KROK 3: Rysowanie danych (Tu zmieniasz typ: .bar, .scatter, .plot)
ax.plot(x, y, color='blue', marker='o', linestyle='--')

# KROK 4: Opis i stylizacja (Najważniejszy krok dla czytelności!)
ax.set_title('Tytuł Twojego Wykresu', fontsize=14)
ax.set_xlabel('Opis osi X')
ax.set_ylabel('Opis osi Y')
ax.grid(True, linestyle=':', alpha=0.6)

# WYŚWIETLENIE
plt.show()
```

1. Zrozum hierarchię (Figure vs Axes)
Wyobraź sobie, że:
- Figure (fig) to cała kartka papieru.
- Axes (ax) to konkretny wykres narysowany na tej kartce.
Dzięki temu zrozumiesz, dlaczego używamy fig, ax = plt.subplots(). To pozwala Ci w przyszłości narysować np. dwa wykresy na jednej kartce.

2. Poznaj "Wielką Trójkę" typów danych
- Nie ucz się wszystkich wykresów naraz. Opanuj te trzy, bo one obsługują 90% przypadków:
- Liniowy (ax.plot): gdy chcesz pokazać trend w czasie (np. zmiana ceny).
- Słupkowy (ax.bar): gdy porównujesz kategorie (np. sprzedaż w różnych działach).
- Punktowy (ax.scatter): gdy szukasz korelacji między dwiema liczbami (np. czy wyższa cena = wyższa jakość).

3. Metoda "Małych Kroków" (Modyfikacja)
- Weź swój działający kod i spróbuj go "popsuć" lub zmienić:
- Zmień kolor słupków na inne kody HEX (np. #FF5733).
- Zmień ax.bar na ax.barh (wykres poziomy).
- Dodaj legendę używając ax.legend(['Grupa A']).

4. Korzystaj z biblioteki Seaborn dla "efektu wow"
- Kiedy poczujesz się pewnie w Matplotlib, sprawdź Seaborn. On używa tego samego szablonu, ale sprawia, że wykresy wyglądają nowocześnie przy mniejszej ilości kodu:

Zadanie 2 – Histogram z KDE
Wygeneruj 1000 losowych liczb z rozkładu normalnego (mean=0, std=1) i stwórz histogram
z nałożoną krzywą KDE.
Wymagania:
- 30 bins
- Kolor: 'steelblue', przezroczystość 0.7
- Dodaj linię pionową dla średniej (czerwona, przerywana)

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Generate 1000 random numbers from a normal distribution (mean=0, std=1)
data = np.random.normal(loc=0, scale=1, size=1000)

# 2. Plot Initialization (Object-Oriented Approach)
fig, ax = plt.subplots(figsize=(6, 4))

# 3. Drawing Histogram with KDE
# Using Seaborn to overlay the Kernel Density Estimate line
sns.histplot(data, bins=30, kde=True, color='steelblue', alpha=0.7, ax=ax)

# 4. Adding a vertical line for the Mean
mean_val = np.mean(data)
ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}')

# 5. Styling and Labels
ax.set_title('Histogram with KDE Curve (Normal Distribution)', fontsize=14)
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
ax.legend()

# 6. Displaying the Plot
plt.show()
```

Zadanie 3 – Wykres słupkowy
Stwórz wykres słupkowy pokazujący średnią cenę biletu na Titanicu według klasy
pasażerskiej.
- Dataset: titanic = sns.load_dataset('titanic')
Wymagania:
- Użyj groupby i mean
- Dodaj wartości na słupkach
- Różne kolory dla klas

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Data Preparation
# Loading the built-in titanic dataset from Seaborn
titanic = sns.load_dataset('titanic')

# Grouping by 'class' and calculating the mean of 'fare'
# We reset the index to keep 'class' as a column for easier plotting
fare_by_class = titanic.groupby('class', observed=False)['fare'].mean().reset_index()

# 2. Plot Initialization
fig, ax = plt.subplots(figsize=(6, 4))

# 3. Drawing the Bar Plot
# Using Seaborn's barplot for easy coloring by class
# palette='viridis' provides different colors for each class
sns_plot = sns.barplot(
    x='class', 
    y='fare', 
    data=fare_by_class, 
    palette='viridis', 
    hue='class', 
    legend=False,
    ax=ax)

# 4. Adding Value Labels on top of each bar (The Requirement)
# We iterate through the bars to place the text
for p in sns_plot.patches:
    sns_plot.annotate(format(p.get_height(), '.2f'),
    (p.get_x() + p.get_width() / 2., p.get_height()),
    ha = 'center', va = 'center',
    xytext = (0, 9),
    textcoords = 'offset points',
    fontsize = 12, fontweight='bold')

# 5. Styling and Labels
ax.set_title('Average Ticket Price (Fare) by Passenger Class', fontsize=14)
ax.set_xlabel('Passenger Class', fontsize=12)
ax.set_ylabel('Average Fare ($)', fontsize=12)
ax.set_ylim(0, fare_by_class['fare'].max() * 1.15) # Adding some space for the labels

# 6. Displaying the Plot
plt.tight_layout()
plt.show()
```

Zadanie 6 – Pie chart
Stwórz wykres kołowy pokazujący proporcje gatunków w datasecie Iris.
Dataset: iris = sns.load_dataset('iris')
Wymagania:
- Użyj value_counts()
- Dodaj procenty na wykresie
- Eksploduj największy kawałek

Dlaczego to zadanie jest ważne w Lekcji 17?
- value_counts(): To jedna z najczęściej używanych funkcji w Pandas. Pozwala szybko zagregować dane kategoryczne przed ich narysowaniem.
- Proportions (Proporcje): Wykres kołowy najlepiej sprawdza się, gdy suma wszystkich elementów wynosi 100%. W datasecie Iris mamy idealnie równe grupy (po 33.3%), co ten wykres doskonale obrazuje.
- Explode (Eksplozja): Ta technika służy do zwrócenia uwagi na konkretny element (np. największy udział w rynku, najczęstszy błąd lub – jak tutaj – konkretny gatunek).

2. Rysowanie wykresu (ax.pie): Kluczowe aspekty techniczne:
- species_counts: To Twoje dane (liczba wystąpień każdego gatunku).
- labels = species_counts.index: Podpisuje każdy kawałek nazwami gatunków (Setosa, Versicolor, Virginica).
- autopct='%1.1f%%': Ten "magiczny" ciąg znaków formatuje liczby tak, aby wyświetlały się jako procenty z jednym miejscem po przecinku.
- 1.1f oznacza: wyświetl liczbę zmiennoprzecinkową z jednym miejscem po przecinku.
- %% na końcu dodaje fizyczny znak % obok liczby.
- startangle = 140: Obraca wykres o 140 stopni. Używa się tego, aby pierwszy kawałek (ten "eksplodujący") znalazł się w ładnym miejscu (np. na górze lub z boku), a nie był przecięty pionową linią startu.
- colors = sns.color_palette('pastel'): Pobiera zestaw łagodnych, pastelowych kolorów z biblioteki Seaborn, dzięki czemu wykres wygląda nowocześnie, a nie jak ze starego Excela.
- explode = explode_values: Przekazuje wcześniej przygotowaną listę wysunięć do wykresu.
- shadow = True: Dodaje pod wykresem delikatny cień, co tworzy efekt 3D i sprawia, że "eksplodujący" kawałek wygląda, jakby unosił się nad kartką.
- startangle: Obraca wykres, co pozwala na lepsze ułożenie etykiet tekstowych.

Wskazówka odnośnie "czystego kodu": Zwróć uwagę, że użyliśmy sns.color_palette('pastel') – to prosty sposób, aby wykres Matplotlib przejął ładną estetykę Seaborn.

explode_values = [0.1 if i == 0 else 0 for i in range(len(species_counts))]
- range(len(species_counts)): Tworzy listę indeksów (0, 1, 2), bo w datasecie Iris mamy 3 gatunki.
- 0.1 if i == 0: Mówi: "Dla pierwszego elementu (indeks 0) ustaw wartość wysunięcia na 0.1".
- else 0: Dla pozostałych elementów ustaw wysunięcie na 0 (zostają na swoim miejscu).
- Efekt: Pierwszy gatunek z listy zostanie lekko "odseparowany" od środka wykresu, co przyciąga wzrok użytkownika.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Data Preparation
# Loading the Iris dataset
iris = sns.load_dataset('iris')
print(iris.head())

# Counting the occurrences of each species
# This fulfills the value_counts() requirement
species_counts = iris['species'].value_counts()
print(species_counts)

# 2. Plot Initialization
fig, ax = plt.subplots(figsize=(6, 6))

# 3. Setting up the "Explode" effect
# In Iris dataset, all species have 50 samples, so we explode the first one
explode_values = [0.1 if i == 0 else 0 for i in range(len(species_counts))]

# 4. Drawing the Pie Chart
# autopct='%1.1f%%' adds percentages to the chart
ax.pie(
    species_counts,
    labels = species_counts.index,
    autopct = '%1.1f%%',
    startangle = 140,
    colors = sns.color_palette('pastel'),
    explode = explode_values,
    shadow = True
)

# 5. Styling
ax.set_title('Proportion of Species in Iris Dataset', fontsize=15)

# 6. Displaying the Plot
plt.tight_layout()
plt.show()
```

- Stylizacja zbiorcza (fig.suptitle): Dodanie głównego tytułu nad całą siatką sprawia, że Twoja praca wygląda jak gotowy raport analityczny.

Zadanie 8 – Heatmap korelacji
Stwórz heatmap korelacji dla zmiennych numerycznych datasetu Tips.
Wymagania:
- Annotacje z wartościami korelacji (2 miejsca po przecinku)
- Paleta kolorów: 'coolwarm'
- Kwadratowe komórki

Dlaczego to zadanie jest kluczowe (Analiza Techniczna):
- select_dtypes(include=['float64', 'int64']): To bardzo ważny krok. Dataset tips zawiera też dane kategoryczne (np. płeć, dzień tygodnia). Próba policzenia korelacji na tekście wyrzuciłaby błąd lub ostrzeżenie. Wybranie tylko liczb to "dobra praktyka", która eliminuje bałagan.
- cmap='coolwarm': Ta paleta jest standardem w nauce. Kolor czerwony oznacza silną korelację dodatnią (1.0), niebieski silną ujemną (-1.0), a biały/szary brak związku (0.0).
- square=True: Dzięki temu wykres jest idealnie proporcjonalny i łatwiejszy do odczytania, niezależnie od liczby analizowanych zmiennych.
- fmt='.2f': Zapewnia precyzję, o którą prosił prowadzący, zachowując jednocześnie czytelność wykresu.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Data Preparation
tips = sns.load_dataset('tips')

# Requirement: Correlation heatmap requires numerical data only
# We select only the columns with numbers to avoid errors
numerical_tips = tips.select_dtypes(include=['float64', 'int64'])
corr_matrix = numerical_tips.corr()

# 2. Plot Initialization
fig, ax = plt.subplots(figsize=(10, 8))

# 3. Drawing the Heatmap
# Requirements: 
# - annot=True: Show correlation values
# - fmt='.2f': 2 decimal places
# - cmap='coolwarm': Blue to Red color palette
# - square=True: Each cell is a square
sns.heatmap(
    corr_matrix, 
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    square=True, 
    linewidths=0.5, 
    ax=ax
)

# 4. Styling and Labels
ax.set_title('Correlation Matrix of Tips Dataset Variables', fontsize=15, pad=20)

# 5. Displaying the Plot
plt.tight_layout()
plt.show()
```

```python
# 1. Data Preparation (Provided by the task)
import pandas as pd
np.random.seed(42)
dates = pd.date_range('2025-01-01', periods=100)
price = 100 + np.cumsum(np.random.randn(100))
df = pd.DataFrame({'date': dates, 'price': price})
print(df.head(3))

# 2. Calculating Moving Averages (MA)
df['MA7'] = df['price'].rolling(window=7).mean()
df['MA21'] = df['price'].rolling(window=21).mean()

# 3. Plot Initialization
fig, ax = plt.subplots(figsize=(14, 7))

# 4. Drawing the Lines
# Main Price Line
ax.plot(df['date'], df['price'], label='Stock Price', color='black', alpha=0.5, linewidth=1)

# Moving Average Lines
ax.plot(df['date'], df['MA7'], label='7-Day MA', color='blue', linewidth=2)
ax.plot(df['date'], df['MA21'], label='21-Day MA', color='red', linewidth=2)

# 5. Shading the area between MA7 and MA21
# This highlights the "gap" or trend strength between the averages
ax.fill_between(df['date'], df['MA7'], df['MA21'], color='gray', alpha=0.3, label='MA Gap')

# 6. Styling and Labels
ax.set_title('Stock Price Analysis: Moving Averages and Trends', fontsize=16, fontweight='bold')
ax.set_xlabel('Date', fontsize=12)
ax.set_ylabel('Price ($)', fontsize=12)
ax.legend(loc='upper right', fontsize=10, frameon=True)


# Adding a grid for better readability
ax.grid(True, linestyle='--', alpha=0.6)

# 7. Displaying the Plot
plt.tight_layout()
plt.show()
```
