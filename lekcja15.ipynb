Zadanie 3 – Iloczyn skalarny i kąt
Oblicz iloczyn skalarny dwóch wektorów a = [1, 2, 3] i b = [4, 5, 6]. Następnie oblicz kąt
między nimi w stopniach.
Wymagania:

Oczekiwany rezultat:
- Iloczyn skalarny (liczba)
- Kąt w stopniach

```python
import numpy as np

# Dataset:
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
# Iloczyn skalarny (dot product)
dot_product = np.dot(a, b) # or a @ b
print(f"a = {a}")
print(f"b = {b}")
print(f"a * b = {dot_product}") # 1*4 + 2*5 + 3*6 = 4 + 10 + 18 = 32

# Ręczne obliczenie dla zrozumienia
manual_dot = (a * b).sum() # Element-wise * potem suma
print(f"\nRęcznie: {manual_dot}")

# Operator @ (preferowany w nowoczesnym kodzie)
dot_with_operator = a @ b
print(f"\nOperator @: {dot_with_operator}")
```

Norma L2: √(Σx²) – długość euklidesowa, "jak daleko od zera"
Norma L1: Σ|x| – suma wartości bezwzględnych, "taxi distance"
Norma L∞: max|x| – największy element (wartość bezwzględna)
Normalizacja: v_unit = v / ||v|| daje wektor o długości 1 w tym samym kierunku

```python
#normy wektorów (L1, L2, L∞)
a = np.array([1, 2, 3])

# Norma L2 (euklidesowa) - najczęstsza
L2 = np.linalg.norm(a)
L2_manual = np.sqrt((a ** 2).sum())

print(f"Vector v: {a}")
print(f"Norma L2 (euklidesowa): {L2:.3f}")
print(f"Ręcznie: √(1² + 2² + 3²) = √(1+4+9) = √14 = {L2_manual:.3f}")

# Norma L1 (Manhattan, taxicab)
L1 = np.linalg.norm(a, ord=1)
L1_manual = np.abs(a).sum()
print(f"\nNorma L1 (Manhattan): {L1:.3f}")
print(f"Ręcznie: |1| + |2| + |3| = {L1_manual}")

# Norma L∞ (maksimum)
Linf = np.linalg.norm(a, ord=np.inf)
Linf_manual = np.abs(a).max()
print(f"\nNorma L∞ (maksimum): {Linf:.3f}")
print(f"Ręcznie: max(|1|, |2|, |3|) = {Linf_manual}")

# Normalizacja wektora do jednostkowej długości (L2=1)
v_normalized = a / L2
norma_normalized = np.linalg.norm(v_normalized)
print(f"\nWektor znormalizowany: {v_normalized}")
print(f"Norma znormalizowanego: {norma_normalized:.10f}") # ≈ 1.0
```

Dot product = suma iloczynów elementów: Σ(a[i] * b[i])
Geometrycznie: a·b = |a||b|cos(θ), gdzie θ to kąt między wektorami
Jeśli a·b > 0 → kąt ostry (wektory "w tym samym kierunku")
Jeśli a·b = 0 → kąt prosty (wektory ortogonalne, niezależne)
Jeśli a·b < 0 → kąt rozwarty (wektory "w przeciwnych kierunkach")

```python
## Geometryczna interpretacja: iloczyn skalarny a długość i kąt
# a · b = |a| * |b| * cos(θ)

norma_a = np.linalg.norm(a) # Długość wektora a
norma_b = np.linalg.norm(b)
cos_theta = dot_product / (norma_a * norma_b)
theta_rad = np.arccos(cos_theta)
theta_deg = np.degrees(theta_rad)

print(f"\n|a| = {norma_a:.3f}")
print(f"|b| = {norma_b:.3f}")
print(f"cos(θ) = {cos_theta:.3f}")
print(f"Kąt θ = {theta_deg:.2f}°")
```

 (tzw. "Zasada Cosinusa"):
 Cosinus = 1 -> Kąt 0 (Identyczne kierunki)
 
 Cosinus = 0 -> Kąt 90 (Wektory prostopadłe, tzw. ortogonalne – nie mają ze sobą nic wspólnego)
 
 Cosinus = -1 -> Kąt 180 (Całkowite przeciwieństwa)

Zadanie 5 – Macierz identyczności i diagonalna
Stwórz macierz identyczności 5×5. Następnie stwórz macierz diagonalną z wartościami [1,
2, 3, 4, 5] na diagonali.
Wymagania:
- Użyj np.eye() i np.diag()
- Wyświetl obie macierze
- Sprawdź, że mnożenie macierzy przez identyczność daje tę samą macierz
Oczekiwany rezultat:
- Macierz identyczności I (5×5)
- Macierz diagonalna D (5×5)
- Weryfikacja: A @ I = A

```python
import numpy as np

# 1. Create a 5x5 Identity Matrix (I)
# # np.eye() creates a matrix with 1s on the diagonal and 0s elsewhere
I = np.eye(5)

# 2. Create a Diagonal Matrix (D)
# np.diag() takes a list and places it on the diagonal of a square matrix
values = [1, 2, 3, 4, 5]
D = np.diag(values)

# Verification: A @ I = A
# matrix D as our "A" for the test
verification = D @ I

# Displaying Results
print("1. Identity Matriź (I):")
print(I)

print("\n2. Diagonal Matrix (D):")
print(D)

print("\n3. Verification (D @ I):")
print(verification)

# Check if they are exactly the same
print("\nIs D @ I equal to D?")
print(np.array_equal(D, verification))
```

Zadanie 7 – Odległość między obserwacjami
Z datasetu Iris wybierz pierwszą i ostatnią obserwację. Oblicz odległość euklidesową i
Manhattan między nimi.
Dataset:
Iris (sklearn.datasets.load_iris)
Wymagania:
- Iris (sklearn.datasets.load_iris)
- Wybierz X[0] i X[-1]
- Oblicz różnicę wektorów
- Oblicz normę L2 różnicy (odległość euklidesowa)
- Oblicz normę L1 różnicy (odległość Manhattan)
Oczekiwany rezultat:
- Dwie wartości odległośc

```python
import numpy as np
from sklearn.datasets import load_iris

# 1. Loading the data
iris = load_iris()
X = iris.data

# 2. Selecting the first and last observation
first_obs = X[0]
last_obs = X[-1]

# 3. Calculating the difference between vectors
# This shows the "distance" in each dimension (feature) separately
diff = first_obs - last_obs

# 4. Calculating Euclidean Distance (L2 Norm of the difference)
dist_euclidean = np.linalg.norm(diff, ord=2)

# 5. Calculating Manhattan Distance (L1 Norm of the difference)
dist_manhattan = np.linalg.norm(diff, ord=1)

# Displaying results
print(f"First observation (X[0]): {first_obs}")
print(f"Last observation (X[-1]): {last_obs}")
print("-" * 40)
print(f"Vector difference: {diff}")
print(f"Euclidean Distance (L2): {dist_euclidean: .4f}")
print(f"Manhattan Distance (L1): {dist_manhattan: .4f}")
```

Zadanie 9 – Standaryzacja datasetu
Wczytaj cały dataset Iris (4 cechy). Zaimplementuj standaryzację (mean=0, std=1) dla
każdej cechy używając operacji NumPy. Porównaj wynik z StandardScaler z sklearn.
Dataset:
- Iris (sklearn.datasets.load_iris)
Wymagania:
- Oblicz mean i std per kolumna ( axis=0 )
- Zastosuj wzór: (X - mean) / std
- Sprawdź mean i std po standaryzacji
- Porównaj z sklearn.preprocessing.StandardScaler
- Oblicz max różnicę między implementacjami
Oczekiwany rezultat:
- Macierz X_standardized (150, 4)
- Weryfikacja: mean≈0, std≈1
- Max różnica vs sklearn

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

iris = load_iris()
X = iris.data

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_standardized = (X - mean) / std

scaler = StandardScaler()
X_sklearn = scaler.fit_transform(X)

max_diff = np.max(np.abs(X_standardized - X_sklearn))

print("--- Macierz X_standardized ---")
print(f"Kształt macierzy: {X_standardized.shape}")
print("Pierwsze 5 wierszy macierzy po standaryzacji:")
print(X_standardized[:5])

print("\nWeryfikacja NumPy (po standaryzacji):")
print(f"Średnie per kolumna: {np.round(np.mean(X_standardized, axis=0), 2)}")
print(f"Odchylenie per kolumna: {np.std(X_standardized, axis=0)}")

print("-" * 40)
print(f"Max różnica między NumPy a sklearn: {max_diff}")
```

Zadanie 10 – Macierz odległości dla k-NN
Wybierz 10 losowych obserwacji z Iris. Oblicz pełną macierz odległości euklidesowych
(10×10) między wszystkimi parami. Wizualizuj jako heatmapę.
Dataset:
- Iris (sklearn.datasets.load_iris)
Wymagania:
- Losowo wybierz 10 obserwacji (użyj np.random.choice na indeksach)
- Użyj euclidean_distances z sklearn lub zagnieżdżone pętle
- Wizualizuj macierz jako heatmapę (seaborn.heatmap lub plt.imshow)
- Diagonala powinna być 0 (odległość od samego siebie)
Oczekiwany rezultat:
- Macierz 10×10
- Heatmapa z annotacjami

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.metrics.pairwise import euclidean_distances

iris = load_iris()
X = iris.data

np.random.seed(42)
random_indices = np.random.choice(X.shape[0], 10, replace=False)
X_subset = X[random_indices]

dist_matrix = euclidean_distances(X_subset, X_subset)

# Wizualizacja jako heatmapa
plt.figure(figsize=(10, 8))
sns.heatmap(dist_matrix, annot=True, cmap='YlGnBu', fmt=".2f")
plt.title("Macierz odległości euklidesowych (10x10) - Podobieństwo próbek")
plt.xlabel("Indeks wybranej próbki")
plt.ylabel("Indeks wybranej próbki")
plt.show()

# Weryfikacja: Sprawdźmy przekątną
print("Przekątna macierzy (odległość od samego siebie):")
print(np.diag(dist_matrix))
```
